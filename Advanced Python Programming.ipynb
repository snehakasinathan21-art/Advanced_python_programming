## 5.4 Challenge to find the factors of a number
 if user enters 20, the output should be "the factors should be 1,2,4,5,10,20"

```python
num = int(input("Enter the number"))
for i in range(1,num+1):
    if num%i ==0:
        print(i)
```

```python
num = int(input("Enter the number"))
result=[]
for i in range(1,num+1):
    if num%i ==0:
        result.append(i)
print("The factors of {} is {}".format(num,result))
```
